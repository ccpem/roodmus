# Creating STAR file with Ground Truth Particle Positions And Defoci

Uses roodmus' simulation/write_starfile utilities:

* read ground truth information from YAML files
* save the dataframe of truth information to file as csv
* use CLI roodmus write_starfile with csv file as input to create a particles.star file
* user particles.star file to extract (or re-extract?) particles to get extracted particle stack
* utilise cryodrgn to train on particle stack in doppio
* interpret latent space representation in another notebook

```python
import os

# read ground truth from yaml files
# data loading for DE-Shaw covid spike partially open set
project_dir = "/mnt/dinaMISMO/synthetic_datasets/MapReconstruction/PreP/extract_frames_6xou_unsteered_0-200ns"
mrc_dir = os.path.join(project_dir, "Micrographs")
figures_dir = os.path.join(project_dir, "figures")

verbose=True

```

```python
# save ground truth dataframe to csv format

# load into dataframe
from roodmus.analysis.utils import load_data

loaded_data = load_data(
    meta_file=None, # no reconstruction file(s)
    config_dir=mrc_dir, # path to roodmus run_parakeet output directory
    particle_diameter=100, # approx
    ugraph_shape=(4000, 4000), 
    ignore_missing_files=False,
    verbose=True,
    enable_tqdm=True,
)

# taking most of this code from src/roodmus/analysis/extract_particles.py?????

```

```python
# make dict into df and save
import pandas as pd

# make df and save to file
truth_csv_fname = os.path.join(project_dir, "truth.csv") 
truth_df = pd.DataFrame(
    loaded_data.results_truth,
    columns=loaded_data.results_truth.keys(),
)
truth_df.to_csv(truth_csv_fname)

# can load with:
"""
pd.read_csv(
    truth_csv_name,
    index_col=0,
)
"""
```
